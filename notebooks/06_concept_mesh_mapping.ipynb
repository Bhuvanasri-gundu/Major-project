# 05 - Concept Extraction & MeSH Mapping

**Objective**: Extract biomedical concepts from MeSH terms and keywords, normalise MeSH terms, build topic-concept mappings.

**Input**:
- `outputs/mhealth_preprocessed.csv`
- `outputs/topic_assignments.csv`

**Outputs**:
- `outputs/document_concepts.csv`
- `outputs/concept_summary.csv`
- `outputs/mesh_normalized.csv`
- `outputs/topic_concept_mapping.csv`

```python
import sys
from pathlib import Path

project_root = Path().resolve()
if project_root.name == 'notebooks':
    project_root = project_root.parent
sys.path.insert(0, str(project_root))
```

```python
import numpy as np
import pandas as pd
from src.config_loader import ensure_dirs
from src.concept_extraction import extract_all_concepts, save_concept_outputs
from src.mesh_normalization import (
    normalize_mesh_terms, build_topic_concept_mapping, save_mesh_outputs,
)

dirs = ensure_dirs()
```

## Load Data

```python
df = pd.read_csv(dirs['outputs'] / 'mhealth_preprocessed.csv', encoding='utf-8-sig')
assignments = pd.read_csv(dirs['outputs'] / 'topic_assignments.csv', encoding='utf-8-sig')
topic_labels = assignments['topic_id'].values
print(f'Records: {len(df):,}')
```

## Extract Concepts

```python
doc_concepts, concept_summary = extract_all_concepts(df, topic_labels=topic_labels)
save_concept_outputs(doc_concepts, concept_summary)
```

## MeSH Normalisation

```python
normalized = normalize_mesh_terms(doc_concepts)
topic_concept_map = build_topic_concept_mapping(normalized)
save_mesh_outputs(normalized, topic_concept_map)
```
